```python
import geopandas
import pandas as pd
```

```python
Kreuzberg_stations_df = pd.read_csv("FilteredDataset/Kreuzberg_stations.csv")
Kreuzberg_stations_df
```

```python
Wilmersdorf_stations_df = pd.read_csv("FilteredDataset/Wilmersdorf_stations.csv")
Wilmersdorf_stations_df
```

```python
line_stops_df = pd.read_csv("PublicTransitDataset/line_stops.csv")
line_stops_df
```

```python
lines_df = pd.read_csv("PublicTransitDataset/lines.csv")
lines_df
```

```python
#print(list(Kreuzberg_stations_df["stop_id"]))
```

```python
Kreuzberg_lines_list = list()
Kreuzberg_lines_set = set()
for i, stop_id in enumerate(line_stops_df["stop_id"]):
    #print(stop_id)
    if stop_id in list(Kreuzberg_stations_df["stop_id"]):
        Kreuzberg_lines_list.append(str(line_stops_df["line_id"][i]))
        Kreuzberg_lines_set.add(str(line_stops_df["line_id"][i]))

```

```python
Kreuzberg_line_ids = list()
Kreuzberg_jahrs = list()
Kreuzberg_linien_names = list()
Kreuzberg_typs = list()
Kreuzberg_dauers = list()
Kreuzberg_ost_wests = list()
Kreuzberg_frequenzs = list()
Kreuzberg_länges = list()
Kreuzberg_starts = list()
Kreuzberg_ziels = list()
for i, line_id in enumerate(lines_df["line_id"]):
    if str(line_id) in Kreuzberg_lines_set:
        Kreuzberg_line_ids.append(lines_df["line_id"][i])
        Kreuzberg_jahrs.append(lines_df["jahr"][i])
        Kreuzberg_linien_names.append(lines_df["linien_name"][i])
        Kreuzberg_typs.append(lines_df["typ"][i])
        Kreuzberg_dauers.append(lines_df["dauer (min)"][i])
        Kreuzberg_ost_wests.append(lines_df["ost_west"][i])
        Kreuzberg_frequenzs.append(lines_df["frequenz"][i])
        Kreuzberg_länges.append(lines_df["länge (km)"][i])
        Kreuzberg_starts.append(lines_df["start"][i])
        Kreuzberg_ziels.append(lines_df["ziel"][i])
                             
```

```python
Kreuzberg_lines_df = pd.DataFrame({'line_id':Kreuzberg_line_ids, 'jahr':Kreuzberg_jahrs, 'linien_name':Kreuzberg_linien_names, 'typ':Kreuzberg_typs, 'dauer (min)':Kreuzberg_dauers, "ost_west":Kreuzberg_ost_wests, "frequenz":Kreuzberg_frequenzs, "länge (km)":Kreuzberg_länges, 'start':Kreuzberg_starts, 'ziel':Kreuzberg_ziels})
Kreuzberg_lines_df
```

```python
Kreuzberg_lines_df.to_csv("FilteredDataset/Kreuzberg_lines.csv")
```
