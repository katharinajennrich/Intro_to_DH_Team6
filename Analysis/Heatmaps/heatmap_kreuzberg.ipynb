Measured by counting for each district, how many stops it includes that are reachable from Kreuzberg

Import Original Dataset: Line Stops

```python
line_stops_df = pd.read_csv("PublicTransitDataset/line_stops.csv")
```

```python
line_stops_df
```

Create base list to fill in Boolean value of connected/not connected to Kreuzberg

```python
line_stops_kreuzberg_boo = [0]*len(line_stops_df["line_id"])
#print(line_stops_kreuzberg_boo)
```

Determine whether each stop is a part of a line that services Kreuzberg

```python

for i, line_id in enumerate(line_stops_df["line_id"]):
    print(line_id)
    if line_id in list(Kreuzberg_lines_df["line_id"]): # change value in list above to match whether each stop is relevant to Kreuzberg
        line_stops_kreuzberg_boo[i]=True 
    else:
        line_stops_kreuzberg_boo[i]=False
```

Add column to line stops dataframe of whether or not the stop is reachable from Kreuzberg as "Kreuzberg_included"

```python
line_stops_df.insert(3,"Kreuzberg_included",line_stops_kreuzberg_boo)
```

```python
line_stops_df
```

Filter dataframe for only those stops connected to Kreuzberg, export

```python
line_stops_df_Kreuzberg = line_stops_df[line_stops_df["Kreuzberg_included"] == True]
```

```python
line_stops_df_Kreuzberg
```

```python
line_stops_df_Kreuzberg.to_csv("FilteredDataset/Kreuzberg_line_stops.csv")
```

Import Original Dataset: all stations
<br> --> will give coordinates with which the district each stop belongs to will be determined

```python
stations_df = pd.read_csv("PublicTransitDataset/stations.csv")
stations_df
```

Create base list for boolean determination of connected to / not connected to Kreuzberg

```python
Kreuzberg_connected_boo = [0]*len(stations_df["stop_id"])
```

determine whether each stop is connected to Kreuzberg

```python
for i, stop_id in enumerate(stations_df["stop_id"]):
    if stop_id in list(line_stops_df_Kreuzberg["stop_id"]):
        Kreuzberg_connected_boo[i] = True
    else:
        Kreuzberg_connected_boo[i] = False
```

Add boolean value to dataframe

```python
stations_df.insert(6,"Kreuzberg_connected",Kreuzberg_connected_boo)
stations_df
```

Filter dataframe for only those stations connected to Kreuzberg

```python
Kreuzberg_connected_stations_df = stations_df[stations_df["Kreuzberg_connected"]==True]
Kreuzberg_connected_stations_df
```

Format co-ordinates in to GeoPandas form

```python
K_latlist = list()
K_longlist = list()
for i, stop in enumerate(Kreuzberg_connected_stations_df["standort"]):
    latlong = stop.split(",")
    K_latlist.append(latlong[0])
    K_longlist.append(latlong[1])
```

```python
K_pointdataframe = pd.DataFrame({'longitude': K_longlist, 'latitude': K_latlist})
K_points = geopandas.points_from_xy(K_pointdataframe.longitude, K_pointdataframe.latitude, crs="EPSG:4326")
K_points = K_points.to_crs("EPSG:4326")
```

Add GeoPandas co-ordinates to dataframe

```python
Kreuzberg_connected_stations_df.insert(5,"Point",K_points)
```

Give the year for each stop by determining it from the stop id, which starts with the year

```python
year = list()
for stop_id in Kreuzberg_connected_stations_df["stop_id"]:
    year.append(str(stop_id)[0:4])
```

Add the year of each stop to the dataframe

```python
Kreuzberg_connected_stations_df.insert(1,"year",year)
Kreuzberg_connected_stations_df
```

```python
Kreuzberg_connected_stations_df.to_csv("AnalysisDataset/Kreuzberg_connected_stations_allyears.csv")
```

Import Berlin Districts Data: this gives the polygons defining each district, allowing each stop reachable from Kreuzberg to be categorized as a part of a district

```python
berlin_gdf = geopandas.read_file("BerlinDistricts.gdb")
berlin_gdf = berlin_gdf.to_crs("EPSG:4326")
berlin_gdf = berlin_gdf.set_index("name")
berlin_gdf
```

Select year

```python
mapyear = 1984
```

Create empty column in district database which will count the occurrences of stops connected to Wilmersdorf for each district, and add to dataframe

```python
num_kreuzberg = [0]*len(berlin_gdf["schlüssel"])
```

```python
berlin_gdf.insert(5,"num_kreuzberg",num_kreuzberg)
```

Filter by year

```python
Kreuzberg_connected_stations_df_mapping = Kreuzberg_connected_stations_df[Kreuzberg_connected_stations_df["year"]==str(mapyear)]
Kreuzberg_connected_stations_df_mapping
```

For each station check if it is in each district, increase num_Kreuzberg value by 1 for the district each stop is in

```python
for i,point in enumerate(Kreuzberg_connected_stations_df_mapping["Point"]): # iterates over all stations connected to Kreuzberg
    for j,district in enumerate(berlin_gdf["schlüssel"]): # iterates over all possible districts in Berlin
        if point.within(berlin_gdf["geometry"][j]): # checks whether the station is in each of the districts
            berlin_gdf["num_kreuzberg"][j] += 1 # adds the station to the district it is within (numerically)
```
